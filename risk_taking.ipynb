```python
# --- MEMBER 1: JAGDEEP SINGH: Data Preparation & Feature Engineering ---


# Working CODE : Implemented Time-Based Non-Equi Join
import pandas as pd
import numpy as np
import statsmodels.api as sm
import statsmodels.formula.api as smf

# Load, clean, and convert time with dayfirst=True to fix the problem
df1 = pd.read_csv('dataset1.csv').rename(columns={'start_time': 'bat_start_time'})
df2 = pd.read_csv('dataset2.csv').rename(columns={'time': 'period_start'})
df1['bat_start_time'] = pd.to_datetime(df1['bat_start_time'], dayfirst=True)
df2['period_start'] = pd.to_datetime(df2['period_start'], dayfirst=True)

# Non-Equi Join Logic
df2['period_end'] = df2['period_start'] + pd.Timedelta(minutes=30)
df_merged = pd.merge(df1, df2, how='cross')
df_merged = df_merged[
    (df_merged['bat_start_time'] >= df_merged['period_start']) &
    (df_merged['bat_start_time'] < df_merged['period_end'])
].copy()
df_merged = df_merged[['bat_landing_to_food', 'risk', 'food_availability', 'rat_minutes', 'season']].dropna()
print(f"M1 Final Result: Successfully merged {len(df_merged)} rows.")

# --- MEMBER 1: JAGDEEP SINGH: Data Preparation & Feature Engineering ---



# Working CODE : Implemented Time-Based Non-Equi Join
import pandas as pd
import numpy as np
import statsmodels.api as sm
import statsmodels.formula.api as smf

# Load, clean, and convert time with dayfirst=True to fix the problem
df1 = pd.read_csv('dataset1.csv').rename(columns={'start_time': 'bat_start_time'})
df2 = pd.read_csv('dataset2.csv').rename(columns={'time': 'period_start'})
df1['bat_start_time'] = pd.to_datetime(df1['bat_start_time'], dayfirst=True)
df2['period_start'] = pd.to_datetime(df2['period_start'], dayfirst=True)

# Non-Equi Join Logic
df2['period_end'] = df2['period_start'] + pd.Timedelta(minutes=30)
df_merged = pd.merge(df1, df2, how='cross')
df_merged = df_merged[
    (df_merged['bat_start_time'] >= df_merged['period_start']) &
    (df_merged['bat_start_time'] < df_merged['period_end'])
].copy()
df_merged = df_merged[['bat_landing_to_food', 'risk', 'food_availability', 'rat_minutes', 'season']].dropna()
print(f"M1 Final Result: Successfully merged {len(df_merged)} rows.")

# --- MEMBER 2: KARANJOT SINGH: Investigation A (OLS Vigilance) ---


#  Working  CODE (Using: Multiple Linear Regression with Control Variable)
print("\n--- 3A. Multiple Linear Regression (Vigilance) ---")

# OLS Vigilance ~ Rat_Intensity + Food_Availability (Control)
formula_A_multi = 'bat_landing_to_food ~ rat_minutes + food_availability'
model_A_multi = smf.ols(formula_A_multi, data=df_merged).fit()
print(model_A_multi.summary())

# Interpretation for Report
print("\nM2 Final Result: Rat minutes is highly significant (P=0.004) even when controlling for food. Predation risk confirmed.")


# --- MEMBER 3: MEHAKPREET SINGH: Investigation A (Logistic Risk-Taking) ---

# Working CODE (Using: GLM with Binomial Link Function)
print("\n--- 3B. Logistic Regression (Risk-Taking) ---")

# Correct Model: GLM with Binomial Family
formula_A_logistic = 'risk ~ rat_minutes'
model_A_logistic = smf.glm(
    formula=formula_A_logistic,
    data=df_merged,
    family=sm.families.Binomial()
).fit()

print(model_A_logistic.summary())

# Interpretation for Report
print("\nM3 Final Result: Log-odds of risk decrease significantly as rat intensity increases (P=0.003).")
```
